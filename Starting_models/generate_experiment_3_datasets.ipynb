# Generate one Experiment 3 numerical dataset
Choose one drag/domain configuration below. Reusing the same seed and physical/wind settings across separate runs gives matched winds and initial states. This notebook does not automatically generate other profiles.

```julia
using Pkg
repository = isdir(joinpath(pwd(),"src")) ? pwd() : dirname(pwd())
Pkg.activate(joinpath(repository,"Starting_models"))
include(joinpath(repository,"src","numerical_dataset.jl"))
using OrdinaryDiffEq
```

## Explicit physical and generation settings

```julia
drag_profile = :uniform                 # :control, :uniform, :interior, :boundary
output_file = joinpath(repository,"Data","experiment_3","$(drag_profile).jld2")
settings = (
    drag_profile=drag_profile,drag_amplitude=1e-4,seed=92026,event_count=18,
    grid_cells=100,domain_length=100.0e3,domain_width=100.0,
    water_density=1000.0,gravity=9.81,chezy_coefficient=50.0,
    minimum_depth=12.0,maximum_depth=28.0,depth_edges=[0.3,0.7],depth_widths=[0.08,0.08],
    start_time=0.0,end_time=72*3600.0,output_cfl=0.7,
    initial_bump_height=1.0,initial_bump_width=0.05,initial_bump_center=0.35,
)
wind_settings = (background_amplitude=(0.6,1.0),background_period_hours=(2.5,4.5),phase=(0.0,2pi),
    storm_amplitude=(1.4,2.2),storm_center_hours=(2.8,3.8),rise_width_hours=(0.45,0.75),
    decay_width_hours=(1.0,1.5),gust_period_minutes=(25.0,65.0),
    second_amplitude_fraction=(0.15,0.35),second_delay_hours=(1.2,1.8),
    second_width_hours=(0.25,0.50),storm_mean_fraction=0.75,storm_gust_fraction=0.25)
solver = Rodas5()
absolute_tolerance = 1e-8
relative_tolerance = 1e-7
nothing
```

## Generate and save the selected dataset

```julia
NumericalDataset.generate_dataset(output_file;settings=settings,wind_settings=wind_settings,
    solver=solver,absolute_tolerance=absolute_tolerance,relative_tolerance=relative_tolerance)
nothing
```

Set `dataset_file` in the main notebook to `output_file`. The saved `solution`, `times`, `grid` and `params` fields are directly compatible with the source data loader. Choose station indices and a pretrained model appropriate for the configured grid.
